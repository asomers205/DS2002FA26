## Git is your lab notebook's time machine

Git records **snapshots** (commits) of your work. GitHub stores them in the cloud and is how you submit in this course.

Core loop you'll use every week:

```bash
git add <files>      # stage what changed
git commit -m "msg"   # snapshot it
git push             # send to GitHub
```

### The three areas

```
working dir  ->  staging (add)  ->  repository (commit)  ->  remote (push)
```

A good commit is small and has a clear message. Examples:
- `Lab 02: fixed broken notebook`
- `HW/Lab 04: pandas challenge`

### Naming matters (course convention)

`YYYY-MM-DD — Topic — Type.ipynb`  where Type ∈ {Lecture, Studio, Lab, Template}.
This sorts correctly everywhere and stays readable.

### Demo: simulate a tiny version history in Python

We can't run `git` in the notebook grader, but we can *model* commits to feel the idea.

````python
commits = []
def commit(message, files):
    commits.append({'id': len(commits) + 1, 'message': message, 'files': files})

commit('Lab 01: setup', ['lab01.ipynb'])
commit('Lab 02: fix broken notebook', ['lab02.ipynb'])
for c in commits:
    print(c['id'], '-', c['message'], c['files'])
````
