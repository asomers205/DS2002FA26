## What is a *data science system*?

A data science **system** is the set of moving parts that turns raw, messy data into a decision:

```
source  ->  ingest  ->  clean  ->  transform  ->  analyze  ->  communicate
```

By hand, you can inspect a few rows. At scale, you need **pipelines**: repeatable steps that do the same thing every time. This course walks up that ladder — databases, Pandas, JSON, APIs, ETL — all inside notebooks.

### Why notebooks, Kaggle/Colab, and Git?

- **Notebooks** mix code, output, and explanation in one document.
- **Kaggle / Colab** give everyone the *same* environment — no 'works on my machine'.
- **Git/GitHub** versions your work and is how you submit.

### Live demo: the smallest possible pipeline

We ingest a tiny dataset, transform it, and produce a decision-ready number.

````python
import pandas as pd

# ingest
sales = pd.DataFrame({
    'game': ['UVA vs A', 'UVA vs B', 'UVA vs C'],
    'ponchos_sold': [120, 15, 340],
    'rain': [True, False, True],
})
sales
````

````python
# transform + analyze: how many ponchos per rainy vs dry game?
sales.groupby('rain')['ponchos_sold'].mean()
````

### The decision

Rainy games move ~10x the ponchos. That single grouped average is a *pipeline output* — and everything in this course is about making that reliable on messy, real data.
