## JSON: the shape of web data

APIs return **JSON** — nested dicts and lists. The skill is turning nesting into a flat, tidy table with `pandas.json_normalize`.

```python
import json, pandas as pd
payload = {
  'game': 'UVA vs Duke',
  'vendors': [
    {'id':'V-01','name':'Hoos Burgers','sales':{'orders':320,'revenue':2400}},
    {'id':'V-10','name':'Cav Merch','sales':{'orders':110,'revenue':2600}},
  ]
}
print(json.dumps(payload, indent=2))
```

### Normalize the nested list

```python
df = pd.json_normalize(payload['vendors'])
df
```

Notice `sales.orders` / `sales.revenue` were flattened with dot-names.
