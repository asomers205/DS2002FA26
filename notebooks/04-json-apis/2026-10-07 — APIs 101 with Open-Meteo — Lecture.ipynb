## Consuming a public API

APIs speak HTTP + JSON. We use **Open-Meteo** (free, no key) to pull real weather — the exact skill your projects need. Pattern: build a URL, `requests.get`, check status, parse `.json()`, normalize.

```python
import requests, pandas as pd

url = 'https://archive-api.open-meteo.com/v1/archive'
params = {
    'latitude': 38.03, 'longitude': -78.51,        # Charlottesville
    'start_date': '2026-10-10', 'end_date': '2026-10-10',
    'hourly': 'temperature_2m,precipitation',
    'timezone': 'America/New_York',
}
try:
    r = requests.get(url, params=params, timeout=30)
    r.raise_for_status()
    data = r.json()
    wx = pd.DataFrame(data['hourly'])
    print(wx.head())
except Exception as e:
    print('API call failed (no internet?):', e)
```

### Always handle failure

Notice the `try/except` and `raise_for_status()`. Networks fail; your pipeline shouldn't crash silently. Note the base URL is cited in a comment — do this in projects.
