## From SQL results to Pandas DataFrames

`pd.read_sql_query` already hands you a DataFrame. But most real data arrives as CSVs. Today: load data into Pandas and do the four moves you'll use constantly — **select, filter, groupby, join**.

```python
import pandas as pd
from io import StringIO

csv = '''vendor,category,item,qty,price
Hoos Burgers,Food,Cheeseburger,3,7.5
Cav Merch,Merch,Foam Finger,1,12.0
Cav Merch,Merch,UVA T-Shirt,2,24.0
Rotunda Tacos,Food,Chicken Tacos,4,6.5
Rally Rain Gear,RainGear,Rain Poncho,10,6.0'''
df = pd.read_csv(StringIO(csv))
df
```

### Select + filter

```python
df[df['category'] == 'Food'][['item', 'qty', 'price']]
```

### A derived column + groupby

```python
df['revenue'] = df['qty'] * df['price']
df.groupby('category')['revenue'].sum().sort_values(ascending=False)
```
