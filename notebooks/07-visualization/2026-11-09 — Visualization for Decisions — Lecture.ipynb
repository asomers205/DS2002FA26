## Visualization for Decisions (not decoration)

A good chart answers a question at a glance. Rules we'll use:
- One chart, one message
- Label axes + title; say the unit
- Sort bars by value
- Don't 3D, don't rainbow, don't chart-junk

```python
import pandas as pd, matplotlib.pyplot as plt
rev = pd.Series({'Food':4200,'Merch':3100,'RainGear':900,'Drink':1800}).sort_values()
ax = rev.plot(kind='barh', title='Revenue by category — 2026-10-10 (USD)')
ax.set_xlabel('revenue (USD)'); plt.tight_layout(); plt.show()
```

### Before/after: a bad chart vs a good one

We'll critique the default and fix it live.

```python
import numpy as np, matplotlib.pyplot as plt
hours = np.arange(9,22)
orders = [10,20,35,60,90,140,220,180,120,80,60,40,25]
fig, ax = plt.subplots()
ax.plot(hours, orders, marker='o')
ax.axvline(15.5, ls='--', color='gray')
ax.annotate('kickoff', (15.5, max(orders)))
ax.set_title('Orders by hour, game day'); ax.set_xlabel('hour'); ax.set_ylabel('orders')
plt.tight_layout(); plt.show()
```
