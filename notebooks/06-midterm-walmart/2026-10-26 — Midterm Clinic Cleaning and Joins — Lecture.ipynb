### Common fix: consolidate fragmented SKUs with a mapping table

```python
import pandas as pd
tx = pd.DataFrame({'sku':['PT-12','SKU#459812','POPTART-STRAW'], 'qty':[40,15,22]})
mapping = pd.DataFrame({'raw_sku':['PT-12','SKU#459812','POPTART-STRAW'],
                        'canonical':['PopTart','PopTart','PopTart']})
tx.merge(mapping, left_on='sku', right_on='raw_sku').groupby('canonical')['qty'].sum()
```

### Common fix: baseline vs surge windows

```python
import pandas as pd, numpy as np
daily = pd.DataFrame({'date': pd.date_range('2024-09-01', periods=14),
                      'orders': [100,110,95,105,120,340,520,480,90,100,110,105,98,102]})
landfall = pd.Timestamp('2024-09-07')
surge = daily[(daily.date >= landfall - pd.Timedelta(days=3)) & (daily.date <= landfall)]
baseline = daily[daily.date < landfall - pd.Timedelta(days=3)]
print('surge avg:', surge.orders.mean(), 'baseline avg:', baseline.orders.mean())
```
