### The Pop-Tarts problem, in miniature

The same product hides under many SKUs and spellings. Consolidating it is the heart of ETL.

```python
import pandas as pd
rows = pd.DataFrame({
  'sku': ['PT-12','SKU#459812','POPTART-STRAW','PT-12'],
  'name': ['pop-tarts strawberry','Strawberry Pop-Tart','Pop-Tarts Strawberry','pop-tarts strawberry'],
  'qty': [40, 15, 22, 40],
})
rows
```

```python
# fragmented view: looks like 3-4 different products
rows.groupby('sku')['qty'].sum()
```

```python
# consolidated view: one product
canon = {'PT-12':'PopTart-Straw','SKU#459812':'PopTart-Straw','POPTART-STRAW':'PopTart-Straw'}
rows['canonical'] = rows['sku'].map(canon)
rows.drop_duplicates().groupby('canonical')['qty'].sum()
```
